# Introduction

Dans le cadre de ce projet, l'objectif est d'exploiter les données commerciales du site de vente de vins **Bottleneck** afin de mieux comprendre les ventes, les prix pratiqués, les niveaux de stock et la rentabilité des produits.

Les données proviennent de plusieurs sources, notamment un fichier ERP contenant des informations sur les produits et les stocks, un fichier relatif aux ventes du site web et une table de liaison permettant de rapprocher les références entre les différentes sources.

L'exploitation de ces données nécessite un travail préalable de contrôle, de nettoyage et de consolidation afin de garantir la cohérence des informations utilisées dans les analyses.

```python
#Calculer la moyenne du prix
mean = df_merge['price'].mean(axis=0, skipna=True, numeric_only=True)
print(mean)
#Calculer l'écart-type du prix
#mesure la dispersion d'un jeu de données par rapport à sa moyenne

ecart_type = df_merge['price'].std()
print(ecart_type)

#Calculer le Z-score
z_score = (df_merge['price'] - mean) / ecart_type
print(z_score)
```

```python
#Quel est le seuil prix dont le z-score est supérieur à 3?
seuil = mean + 3 * ecart_type
print("Seuil prix (z > 3) :", seuil)

z_score_column = df_merge[z_score > 3][['product_id', 'price']]

z_score_column['z_score'] = z_score[z_score > 3].values

print(z_score_column)
```

```python
##############################
# Calculer le CA du site web #
##############################

#Créer une colonne calculant le CA par article
df_merge['ca_par_article'] = df_merge['price'] * df_merge['total_sales']

#Calculer la somme de la colonne "ca_par_article"
#Ce résultat correspond au chiffre d'affaire du site web
ca_total = df_merge['ca_par_article'].sum()
print(ca_total)
```

```python
###############################
# Palmarès des articles en CA #
###############################

#Effectuer le tri dans l'ordre décroissant du CA du dataset df_merge
df_merge = df_merge.sort_values(by="ca_par_article", ascending=False)
#Réinitialiser l'index du dataset par un reset_index
df_merge = df_merge.reset_index(drop=True)
#Afficher les 20 premiers articles en CA
top20 = df_merge.head(20)
print(top20)
```

```python
#Graphique en barre des 20 premiers articles avec plotly express
fig = px.bar(
    top20,
    x="post_title",
    y="ca_par_article",
    title="Top 20 des articles par chiffre d'affaires",
    labels={"post_title": "Article", "ca_par_article": "Chiffre d'affaires"}
)

fig.show()
```

```python
#############################
# Calculer le 20 / 80 en CA #
#############################

#Créer une colonne calculant la part du CA de la ligne dans le dataset
df_merge['part_ca'] = df_merge['ca_par_article'] / df_merge['ca_par_article'].sum()
#Créer une colonne réalisant la somme cumulative de la colonne précedemment créée
df_merge['part_ca_cumulee'] = df_merge['part_ca'].cumsum()
#Grâce aux deux colonnes créées précedemment, calculer le nombre d'articles représentant 80% du CA
nb_articles_80 = (df_merge['part_ca_cumulee'] <= 0.8).sum()
print('Articles représentant 80% du CA:', nb_articles_80)
#Afficher la proportion que représente ce groupe d'articles dans le catalogue entier du site web
proportion = nb_articles_80 / len(df_merge) * 100
print('Part du CA :', proportion.round(2), '%')
```

```python
# Copie du DataFrame pour préserver les données d'origine
df_pareto = df_merge.copy()

# Calcul du chiffre d'affaires par produit
df_pareto["chiffre_affaires"] = (
    df_pareto["price"] * df_pareto["total_sales"]
)

# Vérification des valeurs manquantes
df_pareto = df_pareto.dropna(
    subset=["chiffre_affaires"]
)

# Regroupement par produit
df_pareto = (
    df_pareto
    .groupby("post_title", as_index=False)["chiffre_affaires"]
    .sum()
)
```

```python
# Tri décroissant du chiffre d'affaires
df_pareto = df_pareto.sort_values(
    by="chiffre_affaires",
    ascending=False
).reset_index(drop=True)

# Calcul du chiffre d'affaires cumulé
df_pareto["ca_cumule"] = df_pareto["chiffre_affaires"].cumsum()

# Calcul de la part cumulée du chiffre d'affaires
ca_total = df_pareto["chiffre_affaires"].sum()

df_pareto["part_cumulee"] = (
    df_pareto["ca_cumule"] / ca_total
) * 100

# Affichage des premières lignes
df_pareto.head(10)
```

```python
from matplotlib.ticker import PercentFormatter

# Sélection des 30 premiers produits
top_30 = df_pareto.head(30).copy()

fig, ax1 = plt.subplots(figsize=(15, 7))

# Barres : chiffre d'affaires par produit
ax1.bar(
    range(len(top_30)),
    top_30["chiffre_affaires"]
)

ax1.set_xlabel("Produits classés par chiffre d'affaires décroissant")
ax1.set_ylabel("Chiffre d'affaires")
ax1.tick_params(axis="x", labelbottom=False)

# Courbe : part cumulée du chiffre d'affaires total
ax2 = ax1.twinx()

ax2.plot(
    range(len(top_30)),
    top_30["part_cumulee"],
    marker="o"
)

ax2.axhline(
    y=80,
    linestyle="--",
    label="Seuil de 80 %"
)

ax2.set_ylabel("Part cumulée du chiffre d'affaires (%)")
ax2.yaxis.set_major_formatter(PercentFormatter(xmax=100))
ax2.set_ylim(0, 105)
ax2.legend(loc="upper right")

plt.title(
    "Diagramme de Pareto du chiffre d'affaires — "
    "30 principaux produits"
)

plt.tight_layout()
plt.show()
```

**Interprétation du diagramme de Pareto**

Le diagramme de Pareto permet d'étudier la concentration du chiffre d'affaires entre les différents produits commercialisés par Bottleneck. Les produits sont classés par ordre décroissant de chiffre d'affaires, puis leur contribution cumulée est calculée.

Cette analyse permet d'identifier les produits qui contribuent le plus au chiffre d'affaires et d'évaluer la part du catalogue nécessaire pour atteindre 80 % du chiffre d'affaires total.

Le seuil de 80 % constitue un repère d'analyse et non une règle systématiquement vérifiée. Les résultats doivent être interprétés en tenant compte de la période étudiée et de la qualité des données. Un chiffre d'affaires élevé ne signifie pas nécessairement une marge élevée.

```python
# Nombre de produits nécessaires pour atteindre au moins 80 % du CA
produits_80 = df_pareto[
    df_pareto["part_cumulee"] < 80
]

nombre_produits_80 = len(produits_80) + 1

part_produits = (
    nombre_produits_80 / len(df_pareto)
) * 100

print(f"Nombre total de produits : {len(df_pareto)}")
print(
    f"Nombre de produits nécessaires pour atteindre "
    f"80 % du CA : {nombre_produits_80}"
)
print(
    f"Ces produits représentent {part_produits:.2f} % "
    f"du nombre total de produits."
)
```

```python
#####################################
# Palmarès des articles en quantité #
#####################################

#Effectuer le tri dans l'ordre décroissant de quantités vendues du dataset df_merge
df_merge = df_merge.sort_values(by="total_sales", ascending=False)
#Réinitialiser l'index du dataset par un reset_index
df_merge = df_merge.reset_index(drop=True)
#Afficher les 20 premiers articles en quantité
top20_qt = df_merge.head(20)
print(top20_qt)
```

```python
#############################
# Calculer le 20 / 80 en CA #
#############################

#Créer une colonne calculant la part en quantité de la ligne dans le dataset
df_merge['part_quantite'] = df_merge['total_sales'] / df_merge['total_sales'].sum()
#Créer une colonne réalisant la somme cumulative de la colonne précedemment créée
df_merge['part_quantite_cumulee'] = df_merge['part_quantite'].cumsum()
#Grâce aux deux colonnes créées précedemment, calculer le nombre d'articles représentant 80% des ventes en quantité
nb_articles_80_qt = (df_merge['part_quantite_cumulee'] <= 0.8).sum()
print(nb_articles_80_qt)
#Afficher la proportion que représente ce groupe d'articles dans le catalogue entier du site web
proportion = nb_articles_80_qt / len(df_merge) * 100
print(proportion.round(2), '%')
```

```python
######################################
# Calculer le nombre de mois de stock #
######################################

# 1. Calcul du nombre de mois de stock
# Hypothèse : total_sales correspond aux ventes des 12 derniers mois

df_merge["mois_de_stock"] = np.nan

# On calcule l'indicateur uniquement si les données sont valides
masque_valide = (
    df_merge["stock_quantity"].notna()
    & df_merge["total_sales"].notna()
    & (df_merge["stock_quantity"] >= 0)
    & (df_merge["total_sales"] > 0)
)

df_merge.loc[masque_valide, "mois_de_stock"] = (
    df_merge.loc[masque_valide, "stock_quantity"]
    / (df_merge.loc[masque_valide, "total_sales"] / 12)
)

# 2. Identifier les produits en stock sans ventes
df_sans_ventes = df_merge[
    (df_merge["stock_quantity"] > 0)
    & (df_merge["total_sales"] == 0)
].copy()

print("Produits en stock sans ventes :", len(df_sans_ventes))

# 3. Trier les produits par nombre de mois de stock décroissant
df_sorted_stock = (
    df_merge[df_merge["mois_de_stock"].notna()]
    .sort_values(by="mois_de_stock", ascending=False)
)

# 4. Sélectionner les 20 produits ayant le plus de mois de stock
flop20 = df_sorted_stock.head(20).copy()

# 5. Graphique
fig = px.bar(
    flop20.sort_values("mois_de_stock", ascending=True),
    x="mois_de_stock",
    y="post_title",
    orientation="h",
    title="Top 20 des produits ayant le plus de mois de stock",
    labels={
        "mois_de_stock": "Nombre de mois de stock",
        "post_title": "Produit"
    },
    hover_data=["stock_quantity", "total_sales"]
)

fig.update_layout(
    yaxis={"categoryorder": "total ascending"},
    xaxis_title="Mois de stock estimés",
    yaxis_title="Produit"
)

fig.show()
```

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 6.1 - Utilisation de Great Expectations</h3>
</div>

L'objectif de l'utilisation de Great Expectations est automatiser certains contrôles de qualité afin de détecter les anomalies et de vérifier que les données respectent des règles définies :

- product_id non nul : l'identifiant est nécessaire pour identifier un produit
- product_id unique : Un identifiant ne doit pas correspondre à plusieurs lignes si le modèle de données impose une ligne par produit
- price positif ou nul : Les prix négatifs doivent être examinés
- purchase_price positif ou nul : Les prix d'achat doivent être cohérents avec les règles métier
- stock_quantity positif ou nul : Les stocks négatifs doivent être identifiés et expliqués
- total_sales positif ou nul : Les ventes négatives doivent être examinées selon la définition de l'indicateur

```python
# Note : installation de great_expectations à effectuer si erreur 
import great_expectations as gx

# Initialiser le contexte
context = gx.get_context()

# Créer ou récupérer la suite de tests
suite_name = "suite_qualite_vins"

try:
    suite = context.suites.get(suite_name)

except gx.exceptions.DataContextError:
    suite = gx.ExpectationSuite(name=suite_name)
    suite = context.suites.add(suite)

# Définir les contrôles
expectations = [
    gx.expectations.ExpectColumnValuesToNotBeNull(
        column="product_id"
    ),
    gx.expectations.ExpectColumnValuesToBeUnique(
        column="product_id"
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="price",
        min_value=0
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="purchase_price",
        min_value=0
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="stock_quantity",
        min_value=0
    ),
    gx.expectations.ExpectColumnValuesToBeBetween(
        column="total_sales",
        min_value=0
    ),
]

# Ajouter les règles à la suite
for expectation in expectations:
    suite.add_expectation(expectation)

print(f"{len(suite.expectations)} règles définies.")
```

```python
raw_df = df_merge

raw_df.columns.tolist()
```

```python
# Créer ou récupérer la source de données
datasource_name = "datasource_vins"

try:
    datasource = context.data_sources.get(datasource_name)
except LookupError:
    datasource = context.data_sources.add_pandas(
        name=datasource_name
    )

# Créer ou récupérer l'asset
asset_name = "asset_ventes_vins"

try:
    asset = datasource.get_asset(asset_name)
except LookupError:
    asset = datasource.add_dataframe_asset(
        name=asset_name
    )

# Définir le batch
batch_name = "batch_ventes_vins"

try:
    batch_definition = asset.get_batch_definition(batch_name)
except LookupError:
    batch_definition = asset.add_batch_definition_whole_dataframe(
        batch_name
    )

batch = batch_definition.get_batch(
    batch_parameters={"dataframe": raw_df}
)

# Exécuter les tests
resultat = batch.validate(suite)

print("Validation réussie :", resultat.success)
```
